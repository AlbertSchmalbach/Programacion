###¿Cómo funciona un programa?


Un programa hace que una computadora sea utilizable.

Sin programas, una computadora es solo un objeto inerte.

Las computadoras sólo pueden realizar operaciones muy simples y necesitan instrucciones para resolver tareas complejas.

###Lenguaje: la palabra clave

Imagina que quieres saber la velocidad promedio que alcanzaste durante un viaje largo. Conoces la distancia, conoces el tiempo y necesitas la velocidad.

Naturalmente, la computadora podrá calcularlo, pero no entiende conceptos como distancia, velocidad o tiempo. Por lo tanto, es necesario indicarle que:

1. Acepte un número que represente la distancia.
2. Acepte un número que represente el tiempo de viaje.
3. Divida el primer valor entre el segundo y guarde el resultado en la memoria.
4. Muestre el resultado.

Estas cuatro acciones simples forman un programa.

###Lenguajes naturales vs. lenguajes de programación

Un lenguaje es un medio (y una herramienta) para expresar y registrar pensamientos.

Las computadoras también tienen su propio lenguaje, llamado lenguaje máquina, el cual es muy rudimentario.

Los comandos que reconoce son muy simples. Podemos imaginar que la computadora responde a órdenes como: *toma este número, divídelo entre otro y guarda el resultado*.

> Un conjunto completo de comandos conocidos se llama lista de instrucciones, a veces abreviado como IL.

**Nota:** los lenguajes máquina son desarrollados por humanos.


###Composición de un lenguaje de programación

**Alfabeto:**  Conjunto de *caracteres* permitidos.  
* [A-Z, a-z, 0–10 etc.]

**Léxico [o Diccionario]:**  Conjunto de palabras clave permitidas.
* [`for`, `input`, `while` etc.]

**Sintaxis:** Conjunto de reglas que determina si las palabras tienen sentido.

`for x in range(10):` - es sintacticamente correcto.

`for x in range(10)` – es incorrecto.

> El segundo fragmento de código dará error.

**Semántica**: Conjunto de reglas que determina si las palabras son usadas correctamente.

`my_name = "John"` - semántica y sintácticamente correcto.

`my_name = 47` 	- semánticamente incorrecto pero sintácticamente correcto.

> El segundo fragmento de código no hará lo que esperas, pero hará algo.

###Operadores y su prioridad

**Operadores numéricos** más comunes en Python, junto con un ejemplo para cada uno y el resultado esperado:

| Operador | Descripción                      | Ejemplo en Python | Resultado            |
| -------- | -------------------------------- | ----------------- | -------------------- |
| `+`      | Suma                             | `5 + 3`           | `8`                  |
| `-`      | Resta                            | `10 - 4`          | `6`                  |
| `*`      | Multiplicación                   | `7 * 2`           | `14`                 |
| `/`      | División (flotante)              | `8 / 3`           | `2.6666666666666665` |
| `//`     | División entera (floor division) | `8 // 3`          | `2`                  |
| `%`      | Módulo (resto de la división)    | `8 % 3`           | `2`                  |
| `**`     | Exponenciación                   | `2 ** 3`          | `8`                  |


La **PRIORIDAD** del operador es el **ORDEN** en que se procesan los diferentes operadores. `[Primero – Último]`.

La **ENLACE** del operador es la **DIRECCIÓN** en que se procesan los operadores de la MISMA PRIORIDAD. `[Lado izquierdo o lado derecho]`.

**Tabla de prioridades (precedencia) de operadores en Python**, desde la más alta a la más baja, incluyendo su **binding o Enlace** (asociatividad):

| Prioridad (alta → baja) | Operadores / Forma                                                              | Asociatividad (binding) | Notas / Ejemplos breves                                                  |
| ----------------------- | ------------------------------------------------------------------------------- | ----------------------- | ------------------------------------------------------------------------ |
| 1                       | `()`, `[...]`, `{...}` (agrupación)                                             | —                       | Controlan el orden manualmente.                                          |
| 2                       | `atributo` (`obj.attr`)                                                         | Izq → Der               | Acceso a atributos.                                                      |
| 3                       | `llamada` (`f(...)`), `indexación` (`a[i]`), `slice` (`a[i:j:k]`)               | Izq → Der               | Llamadas, suscripciones y rebanadas.                                     |
| 4                       | `**`                                                                            | **Der → Izq**           | Exponenciación: `2 ** 3 ** 2 == 2 ** (3 ** 2)`.                          |
| 5                       | `+x`, `-x`, `~x` (unarios)                                                      | Der → Izq               | Nota: `-3**2 == -(3**2) == -9` (porque `**` > unario).                   |
| 6                       | `*`, `/`, `//`, `%`                                                             | Izq → Der               | Multiplicación y divisiones.                                             |
| 7                       | `+`, `-`                                                                        | Izq → Der               | Suma y resta.                                                            |
| 8                       | `<<`, `>>`                                                                      | Izq → Der               | Desplazamientos de bits.                                                 |
| 9                       | `&`                                                                             | Izq → Der               | AND a nivel de bits.                                                     |
| 10                      | `^`                                                                             | Izq → Der               | XOR a nivel de bits.                                                     |
| 11                      | `\|`                                                                            | Izq → Der               | OR a nivel de bits.                                                      |
| 12                      | Comparaciones: `<`, `<=`, `>`, `>=`, `!=`, `==`, `in`, `not in`, `is`, `is not` | Especial (encadenables) | `a < b < c` evalúa como `(a < b) and (b < c)` sin evaluar dos veces `b`. |
| 13                      | `not`                                                                           | Der → Izq               | Negación booleana.                                                       |
| 14                      | `and`                                                                           | Izq → Der               | Cortocircuito.                                                           |
| 15                      | `or`                                                                            | Izq → Der               | Cortocircuito.                                                           |
| 16                      | Expresión condicional: `x if cond else y`                                       | Der → Izq               | Ternario.                                                                |
| 17                      | `lambda`                                                                        | Der → Izq               | Crea funciones anónimas.                                                 |

**Notas clave**

* Casi todos los operadores asocian **izquierda a derecha**, excepto `**`, los **unarios** y construcciones como el **ternario** y `lambda`, que asocian **derecha a izquierda**.
* Las **comparaciones** se pueden **encadenar**.
* Los **paréntesis** no son un operador como tal, pero **forzan** el orden de evaluación.

###Ejecuciones condicionales con operadores lógicos



---

#### 📌 1. Concepto básico

En Python, una **ejecución condicional** decide si un bloque de código se ejecuta o no, según si una condición es **verdadera** (`True`) o **falsa** (`False`).

Para combinar o evaluar múltiples condiciones, usamos **operadores lógicos**:

| Operador | Significado                                          | Ejemplo               | Resultado |
| -------- | ---------------------------------------------------- | --------------------- | --------- |
| `and`    | Verdadero si **ambas** condiciones son verdaderas    | `(5 > 2) and (3 < 4)` | `True`    |
| `or`     | Verdadero si **al menos una** condición es verdadera | `(5 > 2) or (3 > 10)` | `True`    |
| `not`    | Invierte el valor lógico de la condición             | `not(5 > 2)`          | `False`   |

---

#### 📌 2. Ejemplo básico

````python
edad = 17
licencia = True

if edad >= 18 and licencia:
    print("Puedes conducir")
else:
    print("No puedes conducir")
````


💡 Aquí se usan **dos condiciones** y el operador `and`. Solo imprime *"Puedes conducir"* si ambas son verdaderas.

---

#### 📌 3. Ejemplos con `or` y `not`

````python
# Uso de 'or'
llueve = False
paraguas = True

if llueve or paraguas:
    print("Puedes salir sin problema")
else:
    print("Mejor espera a que deje de llover")

# Uso de 'not'
esta_apagado = True
if not esta_apagado:
    print("El equipo está encendido")
else:
    print("El equipo está apagado")
````

###Tuplas

---

#### 📌 ¿Qué es una tupla en Python?

Una **tupla** es una colección **ordenada** de elementos **inmutable** (no se puede modificar después de creada).

* Se define usando **paréntesis** `()`
* Puede contener datos de diferentes tipos: enteros, cadenas, booleanos, etc.
* Permite **elementos repetidos**.
* Se accede a sus elementos mediante **índices** (igual que en listas).

💡 **Diferencia clave** con una lista:

* **Lista**: mutable (puede cambiarse).
* **Tupla**: inmutable (no se puede cambiar).

---

#### 📌 Ejemplo básico de tupla

````python
# Crear una tupla
mi_tupla = ("manzana", "banana", "naranja")
print("Crear una tupla:", mi_tupla)

# Acceder a elementos
print("\nAcceder a elementos:")
print("Primer elemento:", mi_tupla[0])   # manzana
print("Último elemento:", mi_tupla[-1]) # naranja

# Longitud de la tupla
print("\nLongitud de la tupla:")
print("Número de elementos:", len(mi_tupla))

# Recorrer la tupla
print("\nRecorrer la tupla:")
for fruta in mi_tupla:
    print("Fruta:", fruta)
````

###Excepciones

---

#### 📌 1. ¿Qué es una excepción?

En Python, una **excepción** es un **error que ocurre durante la ejecución de un programa**.
Cuando ocurre una excepción, el flujo normal del programa se interrumpe, a menos que el error sea **manejado**.

Ejemplos de excepciones comunes:

* `ZeroDivisionError` → dividir entre cero.
* `ValueError` → conversión de tipo inválida.
* `TypeError` → operación o función aplicada a un tipo incorrecto.
* `IndexError` → índice fuera de rango en una lista o tupla.
* `KeyError` → clave inexistente en un diccionario.

---

#### 📌 2. Manejo básico de excepciones (`try` / `except`)

La forma más sencilla de manejar excepciones es con un **bloque try-except**:

````python
try:
    # Código que podría generar un error
    numero = int(input("Ingresa un número: "))
    resultado = 10 / numero
    print("Resultado:", resultado)

except ZeroDivisionError:
    print("Error: No se puede dividir entre cero.")

except ValueError:
    print("Error: Debes ingresar un número válido.")
````

---

#### 📌 3. Usando `else` y `finally`

* **`else`**: se ejecuta si no ocurre ninguna excepción.
* **`finally`**: se ejecuta siempre, ocurra o no una excepción (ideal para liberar recursos).

````python
try:
    numero = int(input("Ingresa un número: "))
    resultado = 10 / numero

except ZeroDivisionError:
    print("Error: División entre cero.")

except ValueError:
    print("Error: Entrada inválida.")

else:
    print("El resultado es:", resultado)

finally:
    print("Fin del programa (bloque finally).")
````

---

#### 📌 4. Capturando cualquier excepción

Si no sabes qué excepción puede ocurrir, puedes capturarla de forma genérica:

````python
try:
    x = int("abc")  # Esto generará un ValueError
except Exception as e:
    print("Ocurrió un error:", e)
````
